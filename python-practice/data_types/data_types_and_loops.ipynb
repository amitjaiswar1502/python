### revenue for 3 quarters ###

```python
revenues = [56 ,60 ,80]
expenses = [20,40,30]
margin_percentage = []

for  i in range(len(revenues)):
    profits = revenues[i] - expenses[i]
    margins = profits * 100/revenues[i]
   
    margin_percentage.append(margins)
    print(f"Margin percentage = {margin_percentage}")
```

```python
financial = [
     {"revenue" : 80 , "expenses" : 40}, #q1
     {"revenue" : 60 , "expenses" : 40}, #q2
     {"revenue" : 70 , "expenses" : 40} #q3
]

for fin in financial:
    profit = fin["revenue"] - fin["expenses"]
    fin['margins'] = profit * 100/fin["revenue"]

financial
```

```python
financials = {
    "Q1" :  {"revenue" : 80 , "expenses" : 24},
    "Q2" :  {"revenue" : 80 , "expenses" : 70},
    "Q3" :  {"revenue" : 80 , "expenses" : 60},
}

for quarters, data in financials.items():
    margin = (data["revenue"] - data["expenses"]) / data["revenue"] * 100
    print(f"{quarters} : {margin:.2f}")

```
